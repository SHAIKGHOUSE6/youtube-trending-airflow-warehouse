```python
import pandas as pd
import numpy as np
```

```python
df = df.drop_duplicates(
    subset=['Video_ID', 'Channel_ID','Date'],
    keep='first'
)
```

```python
df.shape
```

```python
df['PublishedAt'] = pd.to_datetime(
    df['PublishedAt'],
    utc=True
)

df['Published_Date'] = df['PublishedAt'].dt.date
```

```python
df['Published_Month'] = df['PublishedAt'].dt.month
df['Published_Day_Name'] = df['PublishedAt'].dt.day_name()
```

```python
import re

def duration_to_seconds(duration):
    hours = re.search(r'(\d+)H', duration)
    minutes = re.search(r'(\d+)M', duration)
    seconds = re.search(r'(\d+)S', duration)

    h = int(hours.group(1)) if hours else 0
    m = int(minutes.group(1)) if minutes else 0
    s = int(seconds.group(1)) if seconds else 0

    return h * 3600 + m * 60 + s

df['Duration_Seconds'] = df['Duration'].apply(duration_to_seconds)
```

```python
df.sample(5)
```

```python
mega_df = df.drop(columns=['Duration', 'PublishedAt'])
```

```python
mega_df.head()
```

```python
mega_df.info()
```

```python
# amke all column names lowercase 
mega_df.columns = mega_df.columns.str.lower()
```

```python
dim_channel = mega_df[['channel_id', 'channel']].drop_duplicates(
    subset=['channel_id'],keep='first'
).reset_index(drop=True)
```

```python
dim_video = mega_df[
    ['video_id', 'channel_id', 'title', 'published_date',
     'duration_seconds', 'definition', 'caption_available','country_code']
].drop_duplicates(
    subset=['video_id'],
    keep='first'
).reset_index(drop=True)
```

```python
dim_country = mega_df[
    ['country_code', 'country']
].drop_duplicates(
    subset=['country_code'],
    keep='first'
).reset_index(drop=True)

dim_country.insert(
    0,
    'country_id',
    range(1, len(dim_country) + 1)
)
```

```python
mega_df['date'] = pd.to_datetime(mega_df['date'])
mega_df['published_date'] = pd.to_datetime(mega_df['published_date'])

dim_date = mega_df[
    ['date', 'year', 'week','published_date']
].drop_duplicates(
    subset=['date'],
    keep='first'
).reset_index(drop=True)

dim_date['month'] = dim_date['published_date'].dt.month
dim_date['published_day'] = dim_date['published_date'].dt.day
dim_date['published_day_name'] = dim_date['published_date'].dt.day_name()

dim_date.insert(
    0,
    'date_ID',
    range(1, len(dim_date) + 1)
)
```

```python
fact_trending = mega_df[
    ['date', 'video_id', 'views', 'likes', 'comments', 'country_code']
].drop_duplicates(subset=['video_id'], keep='first')
```

```python
# FACT TRENDING
fact_trending = mega_df[
    ['date', 'video_id', 'views', 'likes', 'comments', 'country_code']
].drop_duplicates(
    subset=['video_id', 'country_code', 'date'],
    keep='first'
)
```

```python
print("dim_video:", dim_video.shape)
print("dim_country:", dim_country.shape)
print("dim_date:", dim_date.shape)
print("fact_trending:", fact_trending.shape)
```

```python
dim_video.to_csv('dim_video.csv', index=False)
dim_channel.to_csv('dim_channel.csv', index=False)
dim_country.to_csv('dim_country.csv', index=False)
dim_date.to_csv('dim_date.csv', index=False)
fact_trending.to_csv('fact_trending.csv', index=False)
```

```python
dim_video.to_sql(
    'dim_video',
    engine,
    if_exists='replace',
    index=False
)
dim_country.to_sql(
    'dim_country',
    engine,
    if_exists='replace',
    index=False
)
dim_date.to_sql(
    'dim_date',
    engine,
    if_exists='replace',
    index=False
)
fact_trending.to_sql(
    'fact_trending',
    engine,
    if_exists='replace',
    index=False
)
dim_channel.to_sql(
    'dim_channel',
    engine,
    if_exists='replace',
    index=False
)
```
